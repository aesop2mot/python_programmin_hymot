```python
# 코드 셀 1

import xml.etree.ElementTree as et

# menu.xml 파일을 읽어옵니다. (파일이 업로드되어 있어야 합니다)
tree = et.ElementTree(file='menu.xml')
root = tree.getroot()

print("최상위 태그:", root.tag)
print("타입:", type(root))
```

```python
# 코드 셀 2

# 전체 자식 및 손자 태그 출력
for child in root:
    print('tag:', child.tag, 'attributes:', child.attrib)
    for grandchild in child:
        print('\\ttag:', grandchild.tag, 'attributes:', grandchild.attrib)

print("-" * 30)
# 태그 길이(수) 확인
print("menu의 하위 태그 수:", len(root))
print("breakfast의 item 수:", len(root[0]))
print("root[0]의 타입:", type(root[0]))
```

```python
# 코드 셀 3

# root의 첫 번째 자식 (breakfast) 접근
print("root[0]:", root[0])
print("hours 속성값:", root[0].get("hours"))
print("속성 키 리스트:", root[0].keys())
print("속성 키-값 쌍:", root[0].items())

print("-" * 30)
# tag 이름으로 찾기 (find)
lunch = root.find("lunch")
print("lunch 태그:", lunch)
print("lunch hours 속성값:", lunch.get("hours"))
print("lunch 속성 키:", lunch.keys())
print("lunch 속성 키-값 쌍:", lunch.items())
print("lunch 타입:", type(lunch))
```

```python
# 코드 셀 4

# 같은 이름의 태그가 여러 개일 때 (findall)
print(root[0].findall('item'))

print("-" * 30)
# 태그 안의 내용(text)과 속성(price) 프린트
for item in root[0].findall('item'):
    print("가격:", item.get('price'))
    print("메뉴명:", item.text)
```

```python
# 코드 셀 5

import json

j1 = {"name":"홍길동", "birth":"0525", "age": 30}
print("원본 딕셔너리:", j1)

# json 문자열로 변환
print("json 문자열:", json.dumps(j1))

# 들여쓰기(indent)를 주어 보기 좋게 정렬
print("정렬된 json:\\n", json.dumps(j1, indent=2))

# 리스트와 튜플도 json 문자열로 변경 가능
print("리스트:", json.dumps([1, 2, 3]))
print("튜플:", json.dumps((4, 5, 6)))
```

```python
# 코드 셀 6

# 1. json 문자열을 파이썬 객체로 변경 (loads)
j1 = {"name":"홍길동", "birth":"0525", "age": 30}
d1 = json.dumps(j1)
print("loads 결과:", json.loads(d1))

print("-" * 30)
# 2. 실습용 myinfo.json 파일 임시 생성
with open('myinfo.json', 'w', encoding='utf-8') as f:
    json.dump(j1, f)

# 3. Json 파일 로드 (교안 p.20)
with open('myinfo.json') as f:
    data = json.load(f)
    print("타입:", type(data))
    print("데이터:", data)
```

```python
# 코드 셀 8

# 코랩에는 기본적으로 bs4가 설치되어 있어 바로 import 가능합니다.
from urllib.request import urlopen
from bs4 import BeautifulSoup

html = urlopen("http://pythonscraping.com/pages/page1.html")
bsObj = BeautifulSoup(html.read(), "html.parser")

print(bsObj.h1)
```

```python
# 코드 셀 9

# html 코드를 보기 좋게 정렬하여 출력
print(bsObj.html.prettify())

print("-" * 30)
# 다양한 방식으로 <h1> 태그에 접근
print("h1:", bsObj.h1)
print("html.body.h1:", bsObj.html.body.h1)
print("body.h1:", bsObj.body.h1)
print("html.h1:", bsObj.html.h1)
```
